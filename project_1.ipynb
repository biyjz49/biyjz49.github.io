## Introduction

This project analyzes the U.S. Federal Funds Effective Rate using data from the Federal Reserve.

I will calculate the mean, median, and mode using two methods: pandas and the Python standard library. I will also create a simple visualization to explore changes in interest rates over time.

Data Source: https://fred.stlouisfed.org/series/DFF

## Analysis using pandas

I first use pandas to calculate the mean, median, and mode of the federal funds effective rate.


```python
import pandas as pd

dff = pd.read_csv("DFF.csv")

dff.head()
```

```python
mean_dff = dff["DFF"].mean()
median_dff = dff["DFF"].median()
mode_dff = dff["DFF"].mode().iloc[0]
```

```python
print("Mean:", mean_dff)
print("Median:", median_dff)
print("Mode:", mode_dff)
```

The mean federal funds effective rate is approximately 3.85%, the median is 4.33%, and the mode is 5.33%. Although 5.33% was the most frequently observed rate, the average rate over the entire period was much lower.

## Analysis using the python standard library

I then calculate the same statistics manually using basic Python operations.

I first read the CSV file into a list using basic Python operations

```python
dff_2 = []
with open("DFF.csv", "r") as file:
    for line in file:
        values = line.strip().split(",")
        dff_2.append(values)
```

```python
rates = []

for row in dff_2[1:]:
    rates.append(float(row[1]))
```

To calculate the mean, I add up all the interest rates and divide the total by the number of observations.

To calculate the median, I sort the interest rates in ascending order and find the middle value. If there is an even number of observations, I take the average of the two middle values.

To calculate the mode, I use a dictionary to count how many times each interest rate appears, then identify the rate with the highest frequency.

```python
total = 0

for rate in rates:
    total += rate

mean = total / len(rates)

print("Mean:", mean)

sorted_rates = sorted(rates)
n = len(sorted_rates)
if n % 2 == 0:
    median = (sorted_rates[n // 2 - 1] + sorted_rates[n // 2]) / 2
else:
    median = sorted_rates[n // 2]

print("Median:", median)

counts = {}

for rate in rates:
    if rate in counts:
        counts[rate] += 1
    else:
        counts[rate] = 1

highest_count = max(counts.values())

modes = []

for rate in counts:
    if counts[rate] == highest_count:
        modes.append(rate)

print("Mode:", modes)
```

All three results match previous pandas calculations.

## Data Visualization

The mean, median, and mode summarize interest rates over the entire period, but they do not show how rates changed over time. To explore this, I use pandas to calculate the average interest rate for each year and create a simple text-based bar chart to compare the results.

```python
dff["Year"] = dff["observation_date"].str[:4]
```

```python
print("Average Federal Funds Effective Rate by Year (%)")
print("Each * represents 0.1 percentage points")

print()
for year in sorted(set(dff["Year"])):
    average = dff[dff["Year"] == year]["DFF"].mean()
    stars = round(average * 10)
    average = round(average, 2)
    print(year, ":", "*" * stars, average, "%")
```

The chart shows that the average federal funds effective rate increased dramatically from 2021 to 2023, reaching its highest annual average of 5.14% in 2024. It then declined to 4.21% in 2025 and 3.68% in 2026.

Also note: The data for 2021 and 2026 cover only part of each year.
